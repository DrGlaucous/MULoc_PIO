```python
import numpy as np
import cmath
import math
import pandas as pd
import scipy as sp
import matplotlib.pyplot as plt
import pathlib
import re
from itertools import batched


```

```python

#this doesn't work with ipynb files!
#script_dir = pathlib.Path(__file__).parent.resolve()
input_data_path = "C:\\Users\\EdwardStuckey\\Documents\\GitHub\\MULoc_PIO\\platformIO\\DWM3001CDK-Simple\\visualizations\\cir_data.txt"

def import_data(path: str) -> tuple[list[float], list[float], float]:
    file_p = open(path)

    buffer = file_p.read()

    file_p.close()


    parts = re.split(r'\n', buffer)

    values_str = re.split(r',', parts[0])
    values: list[float] = []
    for a in values_str:
        if(a != ""):
            values.append(float(a))

    values2_str = re.split(r',', parts[1])
    values2: list[float] = []
    for a in values2_str:
        if(a != ""):
            values2.append(float(a))

    first_path = float(parts[2])


    return (values, values2, first_path)



(mag, dir, first_path_pt) = import_data(input_data_path)

fig, axes = plt.subplots(2, 1, figsize=(16, 8))
axes[0].plot(mag, label='CIR Data', color='blue')
axes[0].set_title('CIR Data')
axes[0].set_ylabel('Magnitude')
axes[0].legend()

axes[1].plot(dir, label='CIR Data', color='blue')
axes[1].set_title('CIR Data')
axes[1].set_ylabel('Phase Angle')
axes[1].legend()


print(first_path_pt)

```
